```python
# Discover all available options
import pandas as pd

def explore_pandas_options():
    """Discover and categorize pandas options"""

    # Get all options
    all_options = pd.describe_option()

    # Categories of interest
    categories = {
        'display': 'display.*',
        'computation': 'compute.*',
        'mode': 'mode.*',
        'plotting': 'plotting.*',
        'io': 'io.*'
    }

    for category, pattern in categories.items():
        print(f"\\n=== {category.upper()} OPTIONS ===")
        pd.describe_option(pattern)

# Run the exploration
explore_pandas_options()
```

## 2. get_option() - Retrieving Current Settings

```python
max_rows = pd.get_option('display.max_rows')
print(f"Current max_rows setting: {max_rows}")
```

```python
max_columns = pd.get_option('display.max_columns')  
print(f"Current max_columns setting: {max_columns}")
```

```python
precision = pd.get_option('display.precision')
print(f"Current precision setting: {precision}")
```

```python
# Check mode settings
copy_on_write = pd.get_option('mode.copy_on_write')
print(f"Copy-on-write mode: {copy_on_write}")
```

## 3. set_option() - Modifying Settings

```python
# Before changes
print("=== BEFORE SETTINGS ===")
print(f"Max rows: {pd.get_option('display.max_rows')}")
print(f"Max columns: {pd.get_option('display.max_columns')}")
print(f"Precision: {pd.get_option('display.precision')}")
```

```python
# Set multiple options
pd.set_option('display.max_rows', 50)      # Show up to 50 rows
pd.set_option('display.max_columns', 17)   # Show up to 17 columns
pd.set_option('display.precision', 3)       # Show 3 decimal places
pd.set_option('display.width', 100)         # Set display width
pd.set_option('display.max_colwidth', 30)   # Max column width
```

```python
# After changes
print("=== AFTER SETTINGS ===")
print(f"Max rows: {pd.get_option('display.max_rows')}")
print(f"Max columns: {pd.get_option('display.max_columns')}")
print(f"Precision: {pd.get_option('display.precision')}")
```

```python
# Reset for further examples
pd.reset_option('display.max_rows')
pd.reset_option('display.max_columns')
```
