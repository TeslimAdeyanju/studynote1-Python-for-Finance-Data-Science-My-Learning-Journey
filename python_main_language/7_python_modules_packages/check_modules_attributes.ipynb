### __file__ — Physical Location

```python
import json
```

```python
print(json.__file__)
```

```python
import fda_toolkit as fda
```

```python
print(fda.__file__)
```

```python
# Real-world: Locate your module for debugging
def find_module_location(module):
    return getattr(module, '__file__', 'Built-in module')
```

```python
find_module_location(pd)  # Built-in module
```

### __doc__ — Documentation String

```python
print(fda.__doc__)
```

```python
print(pd.__doc__)
```

```python
import statsmodels.api
```

```python
# Real-world: Auto-generate help text
def show_module_purpose(module):
    doc = getattr(module, '__doc__', 'No documentation available')
    return doc.split('\n')[0] if doc else 'No documentation'

print(show_module_purpose(json))
```

```python
import pandas as pd
```

```python
print(show_module_purpose(pd))
```
