```python
def calculate_average(numbers):
    # Early return for empty list
    if len(numbers) == 0:
        return 0

    # Early return for single item
    if len(numbers) == 1:
        return numbers[0]

    # Normal calculation (final return at end)
    return sum(numbers) / len(numbers)
```

```python
calculate_average([10, 20, 30, 40, 50, 0])
```

```python
def student_state(name, age):
    if age < 18:
        return f"{name} is a minor."
    if age >= 18:
        return f"{name} is an adult."
    if age < 0:
        return "Age cannot be negative."
    else:
        return f"{name} is an adult."
```

```python
student_state("Alice", 14)
student_state("Bob", 20)
student_state("Charlie", -5)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [14, 20, -5]
})
```

```python
df
```

```python
def process_data(df):
    # Step 1: Clean data
    cleaned = [x.strip() for x in df['Name']]

    # Step 2: Filter data
    filtered = [x for x in cleaned if x]

    # Step 3: Transform data
    transformed = [x.upper() for x in filtered]

    # Step 4: Return final result (AT THE END)
    return transformed
```

```python
process_data(df)
```
