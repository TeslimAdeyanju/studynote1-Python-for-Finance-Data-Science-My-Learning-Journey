```python
df
```

```python
df.index
```

```python
# Single Row, Single Column : df.loc[row_label, column_label]
df.loc[10, 'age']
```

```python
# Multiple Rows, Single Column: df.loc[[row_label1, row_label2], column_label]
df.loc[[10, 20, 30], 'age']
```

```python
# 3. Single Row, Multiple Columns: df.loc[row_label, [column_label1, column_label2]]
df.loc[2, ['age', 'fare', 'pclass', 'who']]
```
