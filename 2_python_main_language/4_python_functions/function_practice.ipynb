```python
def calculate_average(numbers):
    # Early return for empty list
    if len(numbers) == 0:
        return 0

    # Early return for single item
    if len(numbers) == 1:
        return numbers[0]

    # Normal calculation (final return at end)
    return sum(numbers) / len(numbers)
```

```python
calculate_average([10, 20, 30, 40, 50, 0])
```

```python
def student_state(name, age):
    if age < 18:
        return f"{name} is a minor."
    if age >= 18:
        return f"{name} is an adult."
    if age < 0:
        return "Age cannot be negative."
    else:
        return f"{name} is an adult."
```

```python
student_state("Alice", 14)
student_state("Bob", 20)
student_state("Charlie", -5)
```

```python
def process_data(df):
    # Step 1: Clean data
    cleaned = [x.strip() for x in df['Name']]

    # Step 2: Filter data
    filtered = [x for x in cleaned if x]

    # Step 3: Transform data
    transformed = [x.upper() for x in filtered]

    # Step 4: Return final result (AT THE END)
    return transformed
```

```python
process_data(df)
```

```python

# Global variable
x = 100

def function1():
    print("Value in 1st function:", x)

def function2():
    global x  # Tells Python to use the global variable x
    x = 5550
    print("Value in 2nd function:", x)

def function3():
    print("Value in 3rd function:", x)

function1()
function2()
function3()

```

```python

def outer_func():
    x = 777

    def inner_func():
        nonlocal x  # Tells Python to use the nearest enclosing variable x
        x = 700  # This creates a new local variable instead of modifying outer x
        print("Value of x inside inner function is:", x)

    inner_func()
    print("Value of x inside outer function is:", x)

# Call the outer function
outer_func()


```

```python
def send_email(to, subject, body, sender = 'info@admin.com', cc = None, bcc = None):
    print(f"Sending email to: {to}")
    print(f"Subject: {subject}")
    print(f"Body: {body}")
    print(f"Sender: {sender}")
    if cc:
        print(f"CC: {cc}")
    if bcc:
        print(f"BCC: {bcc}")
```

```python
send_email("recipient@example.com", "Meeting Reminder", "Don't forget about the meeting at 10 AM tomorrow.")
```
