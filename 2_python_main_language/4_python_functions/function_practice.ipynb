```python
def calculate_average(numbers):
    # Early return for empty list
    if len(numbers) == 0:
        return 0

    # Early return for single item
    if len(numbers) == 1:
        return numbers[0]

    # Normal calculation (final return at end)
    return sum(numbers) / len(numbers)
```

```python
calculate_average([10, 20, 30, 40, 50, 0])
```

```python
def student_state(name, age):
    if age < 18:
        return f"{name} is a minor."
    if age >= 18:
        return f"{name} is an adult."
    if age < 0:
        return "Age cannot be negative."
    else:
        return f"{name} is an adult."
```

## Lambda vs Regular Functions - send_email Example

Lambda functions are designed for **simple, single-expression operations**. The `send_email` function has multiple statements and conditional logic, making it unsuitable for lambda.

```python
# ❌ BAD: Forcing send_email into a lambda (Don't do this!)
# This is technically possible but violates Python best practices

send_email_lambda = lambda to, subject, body, sender='info@admin.com', cc=None, bcc=None: (
    print(f"Sending email to: {to}"),
    print(f"Subject: {subject}"),
    print(f"Body: {body}"),
    print(f"Sender: {sender}"),
    print(f"CC: {cc}") if cc else None,
    print(f"BCC: {bcc}") if bcc else None
)

# This works but is hard to read and maintain
send_email_lambda("test@example.com", "Test", "This is a test")
```

### Why the lambda version is BAD:
1. **Hard to read** - Multiple statements crammed into one line
2. **Hard to debug** - No clear separation of logic
3. **No docstring** - Can't document what it does
4. **Returns a tuple** - The lambda returns (None, None, None...) which is confusing
5. **Against PEP 8** - Python style guide discourages complex lambdas

### ✅ Good Uses of Lambda Functions

```python
# ✅ GOOD: Simple calculations
multiply = lambda x, y: x * y
print(multiply(5, 3))  # Output: 15

# ✅ GOOD: Simple string operations
uppercase = lambda text: text.upper()
print(uppercase("hello"))  # Output: HELLO

# ✅ GOOD: Simple formatting
format_name = lambda first, last: f"{first} {last}".title()
print(format_name("john", "doe"))  # Output: John Doe
```

```python
# ✅ EXCELLENT: Lambda with map, filter, sorted
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print("Squared:", squared)

# Filter even numbers
evens = list(filter(lambda x: x % 2 == 0, numbers))
print("Evens:", evens)

# Sort students by grade
students = [
    {'name': 'Alice', 'grade': 85},
    {'name': 'Bob', 'grade': 92},
    {'name': 'Charlie', 'grade': 78}
]
sorted_students = sorted(students, key=lambda s: s['grade'], reverse=True)
print("Top student:", sorted_students[0]['name'])
```

### 📋 Rule of Thumb: When to use Lambda vs def

**Use regular `def` functions when:**
- ✅ Multiple statements needed
- ✅ Conditional logic (if/else with multiple actions)
- ✅ Multiple print statements
- ✅ Complex logic
- ✅ Need docstrings for documentation
- ✅ Function will be reused in multiple places

**Use lambda when:**
- ✅ Single expression only
- ✅ Simple transformation (x * 2, x.upper(), etc.)
- ✅ Inline function needed (map, filter, sorted)
- ✅ One-time use callback functions
- ✅ Fits on one readable line

```python
# CONCLUSION: For send_email, stick with regular def!
def send_email(to, subject, body, sender='info@admin.com', cc=None, bcc=None):
    """
    Send an email with optional CC and BCC.
    
    Args:
        to: Recipient email address
        subject: Email subject line
        body: Email body text
        sender: Sender email (default: info@admin.com)
        cc: Optional CC recipient
        bcc: Optional BCC recipient
    """
    print(f"Sending email to: {to}")
    print(f"Subject: {subject}")
    print(f"Body: {body}")
    print(f"Sender: {sender}")
    if cc:
        print(f"CC: {cc}")
    if bcc:
        print(f"BCC: {bcc}")

# Much cleaner, more maintainable, and professional!
send_email("recipient@example.com", "Meeting Reminder", "Don't forget about the meeting at 10 AM tomorrow.")
```
