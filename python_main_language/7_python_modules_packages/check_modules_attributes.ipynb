```python
# Real-world: Auto-generate help text
def show_module_purpose(module):
    doc = getattr(module, '__doc__', 'No documentation available')
    return doc.split('\n')[0] if doc else 'No documentation'

print(show_module_purpose(json))
```

```python
import pandas as pd
```

```python
print(show_module_purpose(pd))
```

```python
from email import mime

print(mime.__package__)  # Output: 'email'

# Real-world: Check if module is part of a package
def is_submodule(module):
    return module.__package__ is not None

print(is_submodule(mime))  # True
```

```python
# Real-world: Find what methods are available
def find_public_methods(module):
    return [attr for attr in dir(module) if not attr.startswith('_')]

public_methods = find_public_methods(pd)
print(f"Public methods: {len(public_methods)}")
```

```python
import datetime

namespace = vars(datetime)
print(type(namespace))  # <class 'dict'>
print(len(namespace))   # ~50+ items

# Real-world: Count different types of objects
def analyze_module(module):
    ns = vars(module)
    functions = [k for k, v in ns.items() if callable(v) and not k.startswith('_')]
    classes = [k for k, v in ns.items() if isinstance(v, type) and not k.startswith('_')]
    return {'functions': len(functions), 'classes': len(classes)}

print(analyze_module(datetime))
```

```python
import string

# Safe checking
print(hasattr(string, 'ascii_letters'))  # True
print(hasattr(string, 'nonexistent'))    # False

# Real-world: Conditional imports
def get_function_if_available(module, func_name, default=None):
    if hasattr(module, func_name):
        return getattr(module, func_name)
    return default

sqrt_func = get_function_if_available(math, 'sqrt')
custom_func = get_function_if_available(math, 'custom_func', lambda x: x)
```
