```python
# Example 2: Practical use - creating a unique log file
import datetime

log_name = f"log_{datetime.date.today()}.txt"
try:
    with open(log_name, 'x') as fp:
        fp.write('Log started\n')
except FileExistsError:
    print(f'{log_name} already exists, will append instead')
```

```python
from datetime import datetime

def log_transaction(amount, transaction_type):
    """Add a transaction to the log file."""
    timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')

    with open('transactions.txt', 'a') as fp:
        fp.write(f'{timestamp} | {transaction_type} | {amount}\n')

# Each call adds a new line, keeping all previous transactions
log_transaction(500.00, 'DEPOSIT')
log_transaction(150.00, 'WITHDRAWAL')
log_transaction(2000.00, 'TRANSFER_IN')
```

```python
with open('transactions.txt', 'r') as fp:    print(fp.read())
```

```python
# Example: Write data, then read it back
with open('output.txt', 'w+') as fp:
    # Write some content
    fp.write('Line 1\n')
    fp.write('Line 2\n')
    fp.write('Line 3\n')

    # Move back to the beginning of the file to read
    fp.seek(0)

    # Now read what we just wrote
    content = fp.read()
    print(content)
```

```python
# Example: Add new entry and then show all entries
with open('journal.txt', 'a+') as fp:
    # Add a new entry (always goes to the end)
    fp.write('Day 5: Learned about file modes\n')

    # Move to the beginning to read everything
    fp.seek(0)

    # Read all entries
    all_entries = fp.read()
    print('All journal entries:')
    print(all_entries)
```

```python
def calculate_roi(initial_investment, final_value):
    """Calculate Return on Investment (ROI) as a percentage."""
    if initial_investment == 0:
        return 'Initial investment cannot be zero.'
    
    roi = ((final_value - initial_investment) / initial_investment) * 100
    return f'ROI: {roi:.2f}%'
```

```python
calculate_roi(1000, 1500)
```
